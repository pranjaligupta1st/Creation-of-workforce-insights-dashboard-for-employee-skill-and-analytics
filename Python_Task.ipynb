```python
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Department": [
        "IT", "HR", "Finance", "Sales", "IT",
        "Marketing", "Sales", "HR", "Finance", "IT"
    ],
    "Skill": [
        "Python", "Recruitment", "Excel", "Communication", "SQL",
        "Digital Marketing", "Sales", "Recruitment", "Financial Analysis", "Python"
    ],
    "Experience_Years": [2, 5, 8, 3, 6, 4, 2, 7, 10, 5],
    "Salary": [45000, 55000, 70000, 40000, 65000, 50000, 38000, 68000, 85000, 72000],
    "Performance_Score": [4.2, 3.8, 4.5, 3.5, 4.6, 4.0, 3.2, 4.3, 4.7, 4.4],
    "Job_Satisfaction": [4, 3, 4, 3, 5, 4, 3, 4, 5, 4],
    "Engagement_Score": [4.1, 3.6, 4.3, 3.4, 4.7, 4.0, 3.1, 4.2, 4.6, 4.5]
}

df = pd.DataFrame(data)

df
```

```python
print("Number of Employees:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)
```

## 2. Dataset Exploration

The dataset contains employee information related to department, skills, experience, salary, performance, job satisfaction, and engagement.

The initial exploration helps understand the structure and data types before performing further analysis.

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nNumber of Duplicate Records:")
print(df.duplicated().sum())
```

## 3. Data Quality Check

The dataset was checked for missing values and duplicate records. No missing values or duplicate records were found, so the dataset is suitable for the basic workforce analysis performed in this notebook.

## 4. Employee Skill Analysis

This section analyzes the skills represented in the employee dataset. Understanding the distribution of skills can help organizations identify the capabilities available within their workforce.

```python
skill_count = df["Skill"].value_counts()

print("Employee Skill Distribution:")
print(skill_count)
```

```python
skill_count.plot(kind="bar", figsize=(9, 5))

plt.title("Employee Skill Distribution")
plt.xlabel("Skill")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 5. Department-wise Workforce Analysis

This section examines how employees are distributed across departments. It helps provide a basic understanding of the workforce structure and the number of employees working in each department.

```python
department_count = df["Department"].value_counts()

print("Employees by Department:")
print(department_count)
```

```python
department_count.plot(kind="bar", figsize=(8, 5))

plt.title("Employee Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 6. Salary Analysis

This section analyzes employee salary information. It calculates the average, minimum, and maximum salary and compares average salaries across departments.

```python
average_salary = df["Salary"].mean()
minimum_salary = df["Salary"].min()
maximum_salary = df["Salary"].max()

print("Salary Analysis")
print("-------------------------")
print("Average Salary:", round(average_salary, 2))
print("Minimum Salary:", minimum_salary)
print("Maximum Salary:", maximum_salary)
```

```python
department_salary = df.groupby("Department")["Salary"].mean().sort_values(ascending=False)

print("Average Salary by Department:")
print(department_salary.round(2))
```

```python
department_salary.plot(kind="bar", figsize=(8, 5))

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 7. Employee Performance Analysis

This section analyzes employee performance scores to identify overall performance levels and compare performance across departments.

```python
average_performance = df["Performance_Score"].mean()
highest_performance = df["Performance_Score"].max()
lowest_performance = df["Performance_Score"].min()

print("Performance Analysis")
print("-------------------------")
print("Average Performance Score:", round(average_performance, 2))
print("Highest Performance Score:", highest_performance)
print("Lowest Performance Score:", lowest_performance)
```

```python
department_performance = df.groupby("Department")["Performance_Score"].mean().sort_values(ascending=False)

print("Average Performance by Department:")
print(department_performance.round(2))
```

```python
department_performance.plot(kind="bar", figsize=(8, 5))

plt.title("Average Employee Performance by Department")
plt.xlabel("Department")
plt.ylabel("Average Performance Score")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 8. Job Satisfaction and Employee Engagement

This section analyzes job satisfaction and employee engagement scores. These indicators can help organizations understand how employees experience their workplace and identify differences across departments.

```python
average_satisfaction = df["Job_Satisfaction"].mean()
average_engagement = df["Engagement_Score"].mean()

print("Employee Experience Analysis")
print("-------------------------")
print("Average Job Satisfaction:", round(average_satisfaction, 2))
print("Average Engagement Score:", round(average_engagement, 2))
```

```python
department_experience = df.groupby("Department")[
    ["Job_Satisfaction", "Engagement_Score"]
].mean().round(2)

print("Job Satisfaction and Engagement by Department:")
print(department_experience)
```

```python
department_experience.plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Job Satisfaction and Engagement by Department")
plt.xlabel("Department")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.legend(["Job Satisfaction", "Engagement"])
plt.tight_layout()
plt.show()
```

## 9. High-Performing Employees

This section identifies employees with a performance score of 4.5 or above. This can help demonstrate how workforce data can be used to identify employees with strong performance levels.

```python
high_performers = df[df["Performance_Score"] >= 4.5]

print("High-Performing Employees:")
print(
    high_performers[
        ["Employee_ID", "Department", "Skill", "Performance_Score"]
    ]
)
```

### Interpretation

The analysis identifies employees with relatively high performance scores. Such information can help organizations recognize strong performers and understand the skills and departments associated with higher performance.
